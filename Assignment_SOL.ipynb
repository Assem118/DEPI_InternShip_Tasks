```python
# import all the lib
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error, r2_score
```

```python
# read the dataset using pandas
data = pd.read_csv('/content/Salary_Data.csv')
```

```python
# This displays the top 5 rows of the data
data.head()
```

```python
# Provides some information regarding the columns in the data
data.info()
```

```python
# this describes the basic stat behind the dataset used
data.describe()
```

```python
# These Plots help to explain the values and how they are scattered

# Scatter plot of Years of Experience vs Salary
plt.scatter(data['YearsExperience'], data['Salary'])
plt.xlabel('Years of Experience')
plt.ylabel('Salary')
plt.title('Years of Experience vs Salary')
plt.show()
```

```python
# Cooking the data
X = data['YearsExperience']
X.head()
```

```python
# Cooking the data
y = data['Salary']
y.head()
```

```python
# Create new axis for x column

X = data['YearsExperience'].values.reshape(-1, 1)
```

```python
# Split the data for train and test (70% for training)
X = data[['YearsExperience']]
y = data['Salary']
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Importing Linear Regression model from scikit learn
model = LinearRegression()
```

```python
# Fitting the model
model.fit(X_train, y_train)
```

```python
# Predicting the Salary for the Test values
y_pred = model.predict(X_test)
```

```python
# Plotting the actual and predicted values

plt.figure(figsize=(10, 5))
plt.plot(range(len(y_test)), y_test, color='r', label="Actual")
plt.plot(range(len(y_test)), y_pred, color='b', linestyle='--', label="Predicted")
plt.xlabel('Index')
plt.ylabel('Salary')
plt.title('Actual vs Predicted Salaries')
plt.legend()
plt.show()
```

```python
# plotting the error
plt.plot(range(len(y_test)), y_test - y_pred, color='green')
plt.xlabel('Index')
plt.ylabel('Error')
plt.title('Error Values')
plt.show()
```

```python
# Importing r2_score and mean_squared_error for the evaluation of the model
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)
```

```python
# calculate Mean square error
print(f"Mean Squared Error: {mse}")
```

```python
# Calculate R square vale
print(f"R-Squared Score: {r2}")
```

```python
# Just plot actual and predicted values for more insights
plt.figure(figsize=(10, 5))
plt.scatter(y_test, y_pred, color='r')
plt.xlabel('Actual Salary')
plt.ylabel('Predicted Salary')
plt.title('Actual vs Predicted Values')
plt.show()
```

```python
# Intecept and coeff of the line
print(f'Intercept of the model: {model.intercept_}')
print(f'Coefficient of the line: {model.coef_[0]}')
```

```python
# Equation of the regression line
print(f"Equation of the line: y = {model.intercept_:.2f} + {model.coef_[0]:.2f}x")
```
